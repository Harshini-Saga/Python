```python
#check whether the number is even or odd without using else block
n=int(input())
if n%2==0:
    print(f"{n} is even")
if n%2!=0:
    print(f"{n} is odd")
```

```python
#check whether the number is divisible 
#with 3->return FIZZ
#with 5->return BUZZ
#else->return number
num=int(input())
if num%3==0:
    print("FIZZ")
if num%5==0:
    print("BUZZ")
else:
    print(num)
```

```python
#Check if the num is positive, negative and zero
n=int(input())
if n>0:
    print(f"{n} is Positive")
elif n<0:
    print(f"{n} is Negative")
else:
    print("zero")
```

```python
#If num is divisible by 
#3-FIZZ
#5-BUZZ
#3 and 5-FIZZ BUZZ
#else-num
num=int(input())
if num % 3==0 and num %5==0:
    print("FIZZBUZZ")
elif num%3==0:
    print("FIZZ")
elif num%5==0:
    print("BUZZ")
else:
    print(num)
```
