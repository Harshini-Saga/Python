```python
#2.Memo and Momo
a,b,k=map(int,input().split())
if a%k==0 and b%k==0:
    print("Both")
elif a%k==0:
    print("Memo")
elif b%k==0:
    print("Momo")
else:
    print("No One")
```

```python
#3.Next letter
s=input().lower()
if s=='z':
    print("a")
else:
    s1=ord(s)+1
    print(chr(s1))
```

```python
#4.Ali Baba puzzles
a, b, c, d = map(int, input().split())

if (a + b + c == d or
    a + b - c == d or
    a + b * c == d or
    a - b + c == d or
    a - b - c == d or
    a - b * c == d or
    a * b + c == d or
    a * b - c == d or
    a * b * c == d or
    (a + b) * c == d or
    (a - b) * c == d or
    a * (b + c) == d or
    a * (b - c) == d):
    print("YES")
else:
    print("NO")
```

```python
#5.Sleep Interval
a, b = map(int, input().split())

if a + b > 0 and abs(a - b) <= 1:
    print("YES")
else:
    print("NO")
```
