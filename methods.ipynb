```python
#Access Modifiers
class Account:
    def __init__(self,username,balance,password):
        self.username=username #Public
        self._balance=balance #Protected
        self.__password=password #Private
    def get_balance(self):  #getter method
        return self._balance
    def update_password(self,new_password):   #setter method
         self.__password=new_password
         return "Password is updated"
    def get_password(self):    #getter method
        return self.__password
    def withdraw(self,withdraw_amount):   #setter method
        if self._balance>=withdraw_amount:
            self._balance-=withdraw_amount
            return f"Withdraw Successful and current balance:{self._balance}"
        return "Insufficient balance"
    def deposit(self,deposit_amount):    #setter method
        self._balance+=deposit_amount
        return f"Deposit Successful and current balance:{self._balance}"
    def __total_users(self): #private Method    #getter method
        return "Total users:100"    
user=Account("Harshini",5000,1001)
print(user.get_balance())
print(user.deposit(5000))
print(user.update_password(1002))
print(user.get_password())
#accessing protected members
    #Syntax:obj._attributename
print(user._balance)
#Access private members ny using name mangling
    #Syntax:obj._class__attributename
print(user._Account__total_users())
```
