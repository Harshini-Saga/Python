```python
#1 Recursion
def fun(x):
     if x==0:
         return
     print("I love Recursion")
     return fun(x-1)     
n=int(input())
fun(n)
```

```python
#2 Print from 1 to N
def fun(n):
    if n==0:
        return
    fun(n-1)
    print(n)
n=int(input())
fun(n)
```

```python
#print from n to 1
def print_n_to_1(n):
    if n == 1:
        print(1)
        return
    print(n, end=" ")
    print_n_to_1(n - 1)

n = int(input())
print_n_to_1(n)
```
