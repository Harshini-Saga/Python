```python
#find factors of a number
n=int(input())
l=[]
for i in range(1,n+1):
    if n%i==0:
        l.append(i)
print(l)
if len(l)==2:
    print(f"{n} is prime")
else:
    print(f"{n} is not prime")
```

```python
#finding prime number
n=int(input())
for i in range(2,n//2+1):
    if n%i==0:
        print(f"{n} is not prime")
        break
else:
    print(f"{n} is prime")
```

```python
#find whether the string is palindrome or not
s=input()
r=s[::-1]
if r==s:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
#find whether the string is palindrome or not
s=input().lower()
rev=""
for ch in s:
    rev=ch+rev
if rev==s:
    print(f"{s} is Palindrome")
else:
    print(f"{s} is not a palindrome")  
```

```python
#if num is divisible by 3 square and print sum
n,k=map(int,input().split())
print(sum(i*i for i in range(1,n+1) if i%k==0))
```

```python
#if num is divisible by 3 square and print sum
n,k=map(int,input().split())
summ=0
for i in range(1,n+1):
    if i%k==0:
        n=i*i
        summ=summ+n
print(summ)
```

```python
#Matrix divisible numbers
r=int(input())
nums=[]
for i in range(r):
    nums.append(list(map(int,input().split())))
k=int(input())
res=[]
for i in nums:
    for j in i:
        if j%k==0:
            res.append(j)
print(res)
```

```python
#Number reversing
n=int(input())
rev=0
while n>0:
    digit = n % 10
    rev = rev * 10 + digit
    n //= 10
print(rev)
```

```python
#Move all zeros to the end
lst = [1, 0, 2, 0, 3, 4, 0]
count = lst.count(0)
for i in range(count):
    lst.remove(0)
    lst.append(0)
print(lst)
```

```python
#Perfect number
n = int(input())

s = 0
for i in range(1, n):
    if n % i == 0:
        s += i

if s == n:
    print("Perfect Number")
else:
    print("Not Perfect Number")
```

```python
#Occurences of each word in a sentence
s=input()
l=s.split()
count={}
for i in l:
    if i in count:
        count[i]+=1
    else:
        count[i]=1
print(count)
        
```

```python
seat_type = input().strip()
booking_days = int(input())
festival = input().strip() == "True"
age = int(input())
price = 5000
if seat_type == "Business":
    price += price * 0.40
elif seat_type == "Premium Economy":
    price += price * 0.20
if booking_days > 30:
    price -= price * 0.10
elif booking_days < 7:
    price += price * 0.25
if festival:
    price += price * 0.20
if age >= 60:
    price -= price * 0.15
print(price)
```

```python
age = int(input())
health_score = int(input())
vehicle_type = input().strip()
premium = 10000.0
if age < 25:
    premium *= 1.20
elif age > 50:
    premium *= 1.15
if health_score >= 80:
    premium *= 0.90
elif health_score < 60:
    premium *= 1.20
if vehicle_type == "Sports":
    premium *= 1.30
elif vehicle_type == "SUV":
    premium *= 1.15
print(premium)
```
