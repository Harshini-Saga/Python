```python
#find factors of a number
n=int(input())
l=[]
for i in range(1,n+1):
    if n%i==0:
        l.append(i)
print(l)
if len(l)==2:
    print(f"{n} is prime")
else:
    print(f"{n} is not prime")
```

```python
#finding prime number
n=int(input())
for i in range(2,n//2+1):
    if n%i==0:
        print(f"{n} is not prime")
        break
else:
    print(f"{n} is prime")
```

```python
#find whether the string is palindrome or not
s=input()
r=s[::-1]
if r==s:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
#find whether the string is palindrome or not
s=input().lower()
rev=""
for ch in s:
    rev=ch+rev
if rev==s:
    print(f"{s} is Palindrome")
else:
    print(f"{s} is not a palindrome")  
```

```python
#if num is divisible by 3 square and print sum
n,k=map(int,input().split())
print(sum(i*i for i in range(1,n+1) if i%k==0))
```

```python
#if num is divisible by 3 square and print sum
n,k=map(int,input().split())
summ=0
for i in range(1,n+1):
    if i%k==0:
        n=i*i
        summ=summ+n
print(summ)
```

```python
r=int(input())
nums=[list(map(int,input().split())) for i in range(r)]
k=int(input()) 
res=[num for row in nums for num in row if num%k==0]
print(res)
```

```python
#Matrix divisible numbers
r=int(input())
nums=[]
for i in range(r):
    nums.append(list(map(int,input().split())))
k=int(input())
res=[]
for i in nums:
    for j in i:
        if j%k==0:
            res.append(j)
print(res)
```
