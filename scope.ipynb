```python
#Local Scope
#We can access the local scope variable within the scope, we can't access in global scope
def fun(x):
    x=x+1
    return x
a=10
res=fun(a)
print(res)
```

```python
#Enclosing Scope
def outer(x):
    y=10 #Enclosing Scope
    def inner():
        a=10
        z=y+a #Local Scope
        print(f"This is local scope:{z}")
    inner()
    y=y+x
    print(f"This is Enclosing scope:{y}")
n=5
outer(n)
```

```python
#Enclosing Scope with same variable
def outer(x):
    x=10 #Enclosing Scope
    def inner():
        x=20
        z=x+x #Local Scope
        print(f"This is local scope:{z}")
    inner()
    y=x+x
    print(f"This is Enclosing scope:{y}")
n=5
outer(n)
```

```python
#Global scope
x=10
def outer():
    y=20
    def inner():
        z=30
        res=z+y+x
        print(f"This is local scope:{res}")
    res=y+x
    inner()
    print(f"This is Enclosing scope:{res}")
outer()
```

```python
#Global Scope 
x=10 #Global Scope
def outer():
    y=20 #Enclosing Scope
    def inner():
        z=30 #Local Scope
        res=z+y+x
        print(f"This is local scope:{res}")
    y=y+x
    inner()
    print(f"This is Enclosing scope:{y}")
outer()
```

```python
#Trying to update the global variable in local scope
x=10 #Global Scope
def outer():
    #x=20
    x=x+1
    print(f"This is local scope:{x}")
outer()
print(f"This is global scope:{x}")
```

```python
#Update the global scope variable inside the local scope by using 'global' key
x=10 #Global Scope
def outer():
    global x
    x=x+1
    print(f"This is local scope:{x}")
outer()
print(f"This is global scope:{x}")
```
