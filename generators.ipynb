```python
#generator defination
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res=gen(5)
print(res)
print(next(res))
print(next(res))
print(next(res))
```

```python
#To overcome the stopIteration Error by using loops
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res=gen(5)
for i in res:
    print(i)
```
