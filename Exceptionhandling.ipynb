```python
# Exception Handling blocks
#Without Exception
try:
    a=int(input())
    b=int(input())
    c=a+b
    print(c)
except:
    print("Something Wrong in try block")
else:
    print("There is no exception")
finally:
    print("Program done")
```

```python
# Exception Handling blocks
#With an  Exception
try:
    a=10
    b=0
    c=a/b
    print(c)
except Exception as e:
    print(f"Something Wrong in try block: {e}")
```

```python
# Exception Handling blocks
#Defining multiple exceptions with different except block
try:
    a=10
    b=0
    c=a/b
    print(c)
except ZeroDivisionError as e:
    print(f"Zero Division not possible: {e}")
except IndexError as e:
    print(f"We cannot access element at this position: {e}")
except Exception as e:
    print(f"Something Wrong in try block: {e}")
```

```python
try:
    a=[10,20,30,40]
    c=a[100]
    print(c)
except ZeroDivisionError as e:
    print(f"Zero Division not possible: {e}")
except IndexError as e:
    print(f"We cannot access element at this position: {e}")
except Exception as e:
    print(f"Something Wrong in try block: {e}")
```

```python
try:
    a=10
    c=a.remove(100)
    print(c)
except ZeroDivisionError as e:
    print(f"Zero Division not possible: {e}")
except IndexError as e:
    print(f"We cannot access element at this position: {e}")
except Exception as e:
    print(f"Something Wrong in try block: {e}")
```
