```python
import re
```

```python
#match(pattern,string)
str="The sky is blue and Ocean is blue"
res=re.match("[A-Za-z]+",str)
#print(res)
print(res.span())    #It returns the matched positions
print(res.group())   #It returns the matched value
```

```python
try:
    str="The sky is blue and Ocean is blue"
    res=re.match("z",str)
    print(res)
    print(res.span())    #It returns the matched positions
    print(res.group())   #It returns the matched value
except Exception as e:
    print(f"Something wrong:{e}")
```

```python
try:
    str="The sky is blue and Ocean is blue"
    res=re.search("[A-Za-z]{5}",str)
    print(res.span())    #It returns the matched positions
    print(res.group())   #It returns the matched value
except Exception as e:
    print(f"Something wrong:{e}")
```

```python
#Exact 5 length
try:
    str="The sky is blue and Ocean is blue"
    res=re.search("[A-Za-z]{5}",str)
    print(res.group())   #It returns the matched value
except Exception as e:
    print(f"Something wrong:{e}")
```

```python
#More than length 3
try:
    str="The sky is blue and Ocean is blue"
    res=re.findall("[A-Za-z]{3,}",str)  
    print(res)  
except Exception as e:
    print(f"Something wrong:{e}")
```

```python
#Validate Mobile Numbrs
#All digits,Length should be 10,starts with 6,7,8,9 and remaining should be 0-9
try:
    mobile=input()
    res=re.fullmatch("^[6-9][0-9]{9}",mobile)  
    print(res) 
    if res:
        print(f"{mobile} is valid")
    else:
        print(f"{mobile} is invalid")
except Exception as e:
    print(f"Something wrong:{e}")
```

```python
#Validate pan card number
#Allows upper and digits, length is 10, first 5 are upper next 4 are digits and last is letter
try:
    num=input()
    res=re.fullmatch("^[A-Z]{5}[0-9]{4}[A-Z]{1}",num)  
    if res:
        print(f"{num} is valid")
    else:
        print(f"{num} is invalid")
except Exception as e:
    print(f"Something wrong:{e}")
```

```python
#Checking if the string starts with hello
import re
s=input()
res=re.match("Hello",s)
print(res.span())
print(res.group())
```

```python
#Check if the string ends with .com
import re
s=input()
res=re.search(".com$",s)
print(res.group())
```

```python
#Find all occurrences of the string python
import re
s=input()
res=re.findall("Python",s)
print(res)
print(len(res))
```

```python
#Count how many digits are in the string
import re
s=input()
res=re.findall("\d",s)
print(f"Number Of Digits:{len(res)}")
```

```python
#Find all vowels from the string
s=input()
res=re.findall("[aeiouAEIOU]",s)
print(res)
```

```python
#Check whether the input contains only alphabets
s=input()
if re.fullmatch("[a-zA-Z]",s):
    print("Yes")
else:
    print("No")
```

```python
#Find all spaces in the strinh
s=input()
res=re.findall("\s",s)
print(len(res))
```

```python
#Replaces all spaces with hyphen
s = "Python is easy to learn"
result = re.sub(r"\s", "-", s)
print(result)
```

```python
#Remove all digits from string
s=input()
res=re.sub("\d","",s)
print(res)
```

```python
#Extract all email addresses from the paragraph
import re
s=input()
res=re.fullmatch(r"^[a-z0-9][a-z0-9.]+\@[a-z0-9.]+\.[a-z]{2,3}",s)
print("Valid Email:",res.group())
```

```python
#Extract all hashtags
import re
text = "Learning #Python and #MachineLearning today"
hashtags = re.findall('#\w+', text)
print(hashtags)
```

```python
#Find all dates in Correct format
import re
text = "Dates: 29-08-2026, 01-01-2025"
dates = re.findall("\d{2}-\d{2}-\d{4}", text)
print(dates)
```

```python
#Validate vehicle number
import re
vehicle = "TS09AB1234"
pattern ='^[A-Z]{2}\d{2}[A-Z]{2}\d{4}$'
if re.fullmatch(pattern, vehicle):
    print("Valid Vehicle Number")
else:
    print("Invalid Vehicle Number")
```

```python
#Validate Dates
import re
date=input()
res=re.fullmatch(r"^(0?[1-9]|[12][0-9]|3[01])[-,/](0?[1-9]|1[0-2])[-,/]([0-9]){4}",date)
print(res)
if res:
    print(f"{date} is Valid")
else:
    print(f"{date} is invalid")
```
