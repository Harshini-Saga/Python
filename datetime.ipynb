```python
from datetime import datetime,timedelta
curr_timestamp=datetime.now()  #to get current timestamp
print(curr_timestamp)
today=curr_timestamp.date()   #It extracts date from timestamp
curr_time=curr_timestamp.time() #It extracts time
print(today)
print(curr_time)
print(today.year)
print(today.month)
print(today.day)
print(curr_time.hour)
print(curr_time.minute)
print(curr_time.second)
print(today.strftime("%d-%m-%Y"))
print(curr_time.strftime("%I:%M %p"))
print(today.strftime("%j"))
```

```python
#Timer
import time
from IPython.display import clear_output
sec_counter=0
min_counter=0
hour_counter=0
stop=int(input())
while min_counter!=stop:
    sec_counter+=1
    time.sleep(1)
    if sec_counter==60:
        min_counter+=1
        sec_counter=0
        if min_counter==60:
            hour_counter+=1
            min_counter=0
            if hour_counter==24:
                hour_counter=0
    print(f"{hour_counter}:{min_counter}:{sec_counter}")
    clear_output(wait=True)
```
