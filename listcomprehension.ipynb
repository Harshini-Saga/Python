```python
n=int(input())
l=[]
for i in range(1,n+1):
    l.append(i)
print(l)
```

```python
#1 to n using list comprehension
n=int(input())
lst=[i for i in range(1,n+1)]
print(lst)
```

```python
n=int(input())
lst=[i for i in range(n) if i%2==0]
print(lst)
```

```python
n=int(input())
lst=[2*i for i in range(n)]
print(lst)
```

```python
#return list of integers whether it is even or odd
arr=[10,11,33,54,77,88]
lst=["even" if x%2==0 else "odd" for x in arr]
print(lst)
```

```python
arr=[10,11,33,54,77,88]
lst=["odd" if x&1 else "even" for x in arr]
print(lst)
```

```python
arr=[10,11,33,54,77,88]
lst=[x+1 if x%2==0 else x+0.5 for x in arr]
print(lst)
```

```python
#find one dice combination
print([x for x in range(1,7)])
```

```python
#two dice combination
print([(i,j) for i in range(1,7) for j in range(1,7)])
```
