```python
class Person:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender=gender
    def details(self):
        return f"My Name is {self.name} and I am a {self.age} years old and {self.gender}"
#child class
class Student(Person):
    def __init__(self,name,sid,age):
        self.name=name
        self.sid=sid
        self.age=age
p=Person("Vaishnavi",10,"Female")
s=Student("Harshini",101,21)
print(s.name)
print(p.details())
s.gender="Male"     #adding new instance attribute
print(s.details())
```

```python
#Inherit the parent class constructor in child class by using super ()
#super():This function is used to get the parent class implementation 
class Person:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender=gender
    def details(self):
        return f"My Name is {self.name} and I am a {self.age} years old and {self.gender}"
#child class
class Student(Person):
    def __init__(self,name,sid,age,gender):
        self.sid=sid
        super().__init__(name,age,gender)
    def percentage(self,marks:list):
        return sum(marks)/len(marks)
s=Student("Harshini",101,21,"Female")
print(s.name)
print(s.details())
print(s.percentage([90,80,30]))
```

```python
#Multiple Inheritance
class Father:
    def __init__(self,fname):
        self.fname=fname
    def working(self):
        return "Father working as farmer"
class Mother:
    def __init__(self,mname):
        self.mname=mname
    def working(self):
        return "Mother working as housewife"
class Child(Father,Mother):
    def __init__(self,fname,mname,name):
        self.name=name
        Father.__init__(self,fname)
        Mother.__init__(self,mname)
    def working(self):
        f=Father(self.fname)
        m=Mother(self.mname)
        print(f.working())
        print(m.working())
c=Child("Gangadhar","Pushpa","Lasya")
print(c.fname,c.mname,c.name)
c.working()  #Based on Method Resolution order
```

```python
#Multiple Inheritance
class Father:
    def __init__(self,fname):
        self.fname=fname
    def working1(self):
        return "Father working as farmer"
class Mother:
    def __init__(self,mname):
        self.mname=mname
    def working2(self):
        return "Mother working as housewife"
class Child(Father,Mother):
    def __init__(self,fname,mname,name):
        self.name=name
        Father.__init__(self,fname)
        Mother.__init__(self,mname)
    def working(self):
        print(super().working1())
        print(super().working2())
c=Child("Gangadhar","Pushpa","Lasya")
print(c.fname,c.mname,c.name)
c.working()
```

```python
#Hierarchial Inheritance
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def details(self):
        return f"My Name is {self.name} and I am a {self.age} years old"
class Employee(Person):
    def __init__(self,name,age,salary):
        self.salary=salary
        super().__init__(name,age)
    def emp_details(self):
        return f"Salary:{self.salary}"
class Student(Person):
    def __init__(self,name,age,sid):
        self.sid=sid
        super().__init__(name,age)
    def stu_details(self):
        return f"Roll No:{self.sid}"
s=Student("Harshini",21,101)
e=Employee("Vaishnavi",25,20000)
print(s.stu_details())
print(e.emp_details())
print(s.details())
print(e.details())
```

```python
#Hybrid Inheritance
class Person:
    def __init__(self,name):
        self.name=name
    def details(self):
        return f"my name is {self.name}"
class Student(Person):
    def study(self):
        return f"{self.name} is studying"
class Employee(Person):
    def work(self):
        return f"{self.name} is working"
class Intern(Student,Employee):
    def intern(self):
        return f"{self.name} is doing an internship"
e=Intern("Harshini")
print(e.details())
print(e.study())
print(e.work())
print(e.intern())

    
```
