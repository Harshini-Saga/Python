```python
#
name="harshini"
age=20
print("My name is " + name + " and i am " + str(age) + "Years old") 
```

```python
##Output Formatting
#1.Old Style Formatting(using commma):
#
name="harshini"
age=20
print("My name is",name,"and i am",str(age),"Years old") 
```

```python
#2.Access Specifiers
# integer-%d
# float-%f
# string-%s
name="harshini"
age=20
print("My name is %s and i am %d Years old"%(name,age))
```

```python
#3.Dot Format Method
#
name="harshini"
age=20
print("My name is {} and i am {} Years old".format(name,age)) 
```

```python
#4.f-Format
#
name="harshini"
age="20"
percentage=98.2568
print(f"My name is {name} and i am {age} Years old and my percentage is {percentage}")
print(f"My name is {name} and i am {age} Years old and my percentage is {percentage:.2f}")
```
