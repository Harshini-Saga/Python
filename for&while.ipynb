```python
#Remove target element using for loop
l=list(map(int,input().split()))
target=int(input())
l1=[]
for i in l:
    if i!=target:
        l1.append(i)
print(l1)
    
```

```python
#remove target element using for range and index
l=list(map(int,input().split()))
target=int(input())
l1=[]
n=len(l)
for i in range(n):
    if l[i]!=target:
        l1.append(l[i])
print(l1)
```

```python
#remove target element using while loop
arr=list(map(int,input().split()))
target=int(input())
res=[]
n=len(arr)
i=0
while i<n:
    if arr[i] != target:
        res.append(arr[i])
    i+=1
print(res)
```

```python
arr=list(map(int,input().split()))
target=int(input())
n=len(arr)
i=0
while i<n:
    if arr[i]==target:
        arr.pop(i)
        n-=1
    else:
        i+=1
print(arr)
    
```

```python
#Index of target element
arr=list(map(int,input().split()))
target=int(input())
n=len(arr)
for i in range(n):
    if arr[i]==target:
        print(i)
else:
    print(-1)
```

```python
#Index of target element
arr=list(map(int,input().split()))
target=int(input())
n=len(arr)
for i in range(n):
    if arr[i]==target:
        print(i)
        break
else:
    print(-1)
```
