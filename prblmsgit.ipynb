```python
#4.Count how many even and odd numbers are present
n=int(input())
arr=list(map(int,input().split()))
even,odd=0,0
for i in arr:
    if i%2==0:
       even+=1
    else:
        odd+=1
print(f"Even:{even}")
print(f"Odd:{odd}")
```

```python
#7.Count positive, negative and zero
n=int(input())
arr=list(map(int,input().split()))
pos,neg,z=0,0,0
for i in arr:
    if i>0:
        pos+=1
    elif i<0:
        neg+=1
    else:
        z+=1
print(f"Positive:{pos}")
print(f"Negative:{neg}")
print(f"Zero:{z}")
```

```python
#Find duplicate elements
n=int(input())
arr=list(map(int,input().split()))
for i in range(n):
    count=0
    for j in range(n):
        if arr[i]==arr[j]:
            count+=1
    if count>1 and arr[i] not in arr[:i]:
        print(arr[i],end=" ")
```

```python
#Longest word
s = input()
words = s.split()
longest = words[0]
for word in words:
    if len(word) > len(longest):
        longest = word
print(longest)
```

```python
#Toggle Case
s = input()
result = ""
for ch in s:
    if 'a' <= ch <= 'z':
        result += chr(ord(ch) - 32)
    elif 'A' <= ch <= 'Z':
        result += chr(ord(ch) + 32)
    else:
       result += ch
print(result)
```
