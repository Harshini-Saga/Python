```python
#Check whether the number is positive or not
num=int(input())
if num>=0:
    print(f"{num} is Positive")
else:
    print(f"{num} is Negative")
```

```python
#check whether number is divisible by 3
num=int(input())
if num%3 == 0:
    print(f"{num} is divisble by 3")
else:
    print(f"{num} is not divisble by 3")
```

```python
#Check whether num is divisible by 5 and 3
num=int(input())
if num%3==0 and num%5==0:
    print(f"{num} is divisible by 5 and 3")
else:
    print(f"{num} is not divisible by 5 and 3")
```

```python
#Check whether the number is perfect square or 
n=int(input())
val=n**0.5
print(val)
if val==int(val):
    print(f"{n} is Perfect square")
else:
    print(f"{n} is Not a perfect square")
```

```python
#return index of target if target exist in list otherwise -1
n=list(map(int,input().split()))
t=int(input())
if t in n:
    print(n.index(t))
else:
    print(-1)
```
