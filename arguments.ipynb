```python
#Type of Arguments
#1.Positional Arguments:The arguments are assigns to parameters as per the positional order
def sub(x:int,y:int)->int:
    c=x-y
    return c
#Function callling block
a,b=10,20
res=sub(a,b)
print(res)
print(sub(b,a))
```

```python
#Type of Arguments
#2.Keyword Arguments:Here arguments are passed along with the parameters
def sub(x:int,y:int)->int:
    c=x-y
    return c
#Function callling block
a,b=10,20
print(sub(x=a,y=b))
print(sub(y=b,x=a))
```

```python
#Sum of two numbers,Sumof three numbers,Sum of four numbers
def summ(x,y):
    c=x+y
    return c
def sum1(x,y,z):
    res=x+y+z
    return res
def sum2(x,y,z,w):
    res1=x+y+z+w
    return res1
x,y,z,w=map(int,input().split())
print(summ(x,y))
print(sum1(x,y,z))
print(sum2(x,y,z,w))
```

```python
#3.Default Arguments:The parameters defined with some default values, and it assign the argument value
#f you pass the values for default parameters otherwise, it will assign default values

def add(a,b,c=0,d=0):
    res=a+b+c+d
    return res
print(add(10,20))
print(add(10,20,5))
print(add(10,20,5,3))
```

```python
def mul(a,b,c=1,d=1):
    res=a*b*c*d
    return res
print(mul(10,20))
print(mul(10,20,5))
print(mul(10,20,5,3))
```

```python
#4.Variable length of positional arguments
def add(a,b=0,*args):
    s=0
    for num in args:
        s+=num
    return s+a+b
print(add(1))
print(add(1,2,3,4,5,5,3,1))
print(add(1,2,3,4,5,6,7,8,9))      
```

```python
#5.variable length of keyword arguments
def details(name,age,*marks,**kwargs):
    print(type(kwargs))
    print(kwargs)
    print(marks)
print(details("ravi",20,85,35,75,gender="Male",location="India"))
```
