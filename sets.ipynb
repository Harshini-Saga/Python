#### Set Methods
- Adding element to set
    - add(ele):It adds element to set if element not exists in set, otherwise nothing.
    - update(iterable):It adds all iterable elements to set
- Remove elements from set
    - Remove(ele):It removes element from set if element exists in set otherwise 'KeyError'.
    - discard(ele):It removes the element from set if element exists in set otherwise nothing.
    - pop():it remove and returns the element from set 
- Set Operations
    - union(another_sets):It returns the all common elements from all sets.
    - intersection(another_sets):It returns the common elements from the all sets.
    - difference(another set):It returns the set elements those are not lies in another sets. 
    - symmetric differenece(another_sets):It returns the Non-common elements from all sets.
    
    




```python
#Add,Remove and discard
s1={10,20,40,78,90}
s1.add(30)
print(s1)
s1.remove(20)
print(s1)
s1.discard(78)
print(s1)
s1.remove(40)
print(s1)
s1.discard(50)
print(s1)
```

```python
#Set Operations
s1={1,2,3,4,6,9}
s2={5,6,7,8,3}
s3={2,5,3,10}
print(s1.union(s2))
print(s1.union(s2,s3))
print(s1.intersection(s2))
print(s1.intersection(s2,s3))
print(s1.difference(s2))
print(s1.difference(s2,s3))
print(s1.symmetric_difference(s2))
print(s1.symmetric_difference(s2,s3))#Type Error
```

```python
s1={1,2,3,4,6,9}
s2={5,6,7,8,3}
s3={2,5,3,10}
print(s1|(s2))#Union
print(s1|s2|s3)#Union
print(s1&(s2))#Intersection
print(s1& s2 & s3)#intersection
print(s1-(s2))#Difference
print(s1-s2-s3)#Difference
print(s1^(s2))#SymmetricDifference
```
