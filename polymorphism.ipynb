```python
#Method Overriding
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def details(self):
        return f"My name is {self.name} and I am {self.age} years old"
class Student(Person):
    def __init__(self,name,age,rollno):
        self.rollno=rollno
        super().__init__(name,age)
    def details(self):
        return "I am student"
s=Student("harshini",21,101)
p=Person("lasya",16)
print(p.details())
print(s.details())
```

```python
#Method overloading
class math:
    def add(self,x,y):
        return x+y
    def add(self,x,y,z):
        return x+y+z
    def add(self,a,b,c,d):
        return a+b+c+d
m=math()
print(m.add(1,2,3,4))
#print(m.add(1,2,3))
#print(m.add(1,2))
```

```python
#Python directly doesnot support method overloading,but it can acheive through default or arbitary arguments
class math:
    def add(self,a,b,c=0,d=0):
        return a+b+c+d
m=math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))    
```

```python
#Relational operators
class Point:
    def __init__(self,x):
        self.x=x
    def __gt__(self,other):
        return self.x>other.x
    def __lt__(self,other):
        return self.x<other.x
    def __ge__(self,other):
        return self.x>=other.x
    def __le__(self,other):
        return self.x<=other.x
    def __ne__(self,other):
        return self.x!=other.x
    def __eq__(self,other):
        return self.x==other.x
p1=Point(20)
p2=Point(10)
print(p1>p2)
print(p1<p2)
print(p1>=p2)
print(p1<=p2)
print(p1!=p2)
print(p1==p2)
```

```python
#Arithmetic Operators
class Point:
    def __init__(self,x):
        self.x=x
    def __add__(self,other):
        return self.x+other.x
    def __sub__(self,other):
        return self.x-other.x
    def __mul__(self,other):
        return self.x*other.x
    def __truediv__(self,other):
        return self.x/other.x
    def __floordiv__(self,other):
        return self.x//other.x
    def __mod__(self,other):
        return self.x%other.x
    def __pow__(self,other):
        return self.x**other.x
p1=Point(10)
p2=Point(5)
print(f"Addition:{p1+p2}")
print(f"Subtarction:{p1-p2}")
print(f"Multiplication:{p1*p2}")
print(f"Division:{p1/p2}")
print(f"Floor Division:{p1//p2}")
print(f"Modulus:{p1%p2}")
print(f"Power:{p1**p2}")
```
