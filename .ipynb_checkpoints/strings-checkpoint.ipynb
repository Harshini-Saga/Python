```python
#split()
string="The ocean is blue and sky is BLUE"
print(string.split())
```

```python
string="The ocean is blue and sky is BLUE 12344@"
print(string.split('is'))
print(string.split('Z'))
print(string.split('blue'))
print(string.split('@'))
```

```python
#Index Finding Methods
#index(sub_string):It returns the first occurance index of substring if exists, otherwise "ValueError"
#find(sub_string):It returns the first occurance index of substring if exists, otherwise "-1"
string = "The ocean is blue and sky is blue"
print(string.index("o"))
print(string.index("is"))
print(string.index("is",14))#It starts checking from 14th index
print(string.index("Z"))
```

```python
#find(sub_string):It returns the first occurance index of substring if exists, otherwise "-1"
string = "The ocean is blue and sky is blue"
print(string.find("o"))
print(string.find("is"))
print(string.find("is",14))#It starts checking from 14th index
print(string.find("Z"))
```

```python
#replace(old,new):It replaces the old string with new string in string
string = "The ocean is blue and sky is blue"
print(string.replace('blue','red'))
print(string.replace('blue','red',1))#It replaces only 1 occurrance
```

```python
#"sep".join(iterable):It joins the all iterable elements with separator
lst=['The','ocean','is','blue','and','sky','is','BLUE']
print("".join(lst))
print(" ".join(lst))
print("_".join(lst))
```

```python
#string checking methods
-islower():It returns true when entire string is in lower case
-isupper():It returns true when entire string is in upper case
-istitle():It returns true when string is in title format
-isalpha():It returns true when string contains alphabets(lowercase and uppercase)
-isalnum():It returns true when string contains either alphabets or digits
-isdigit():It returns true when entire string contains only digits
-startswith():It returns true when string starts with sub_string
-endswith():It returns true when string ends with sub_string
```

print("hello".islower())
print("Hello".isupper())
print("HELLO".istitle())
print('1234'.isdigit())
print('abcABC'.isalnum())
print("Hello World")
