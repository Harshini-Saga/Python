```python
#Define set
#Set is unordered collection
s={10,2,4,9,3,"Code","Hello",2.25}
print(s)
print(hash(10))
print(hash("Code"))
print(hash(2.25))
```

```python
#Set allows only unique elements
s={1,3,"Harshini","Codegnan",1,3}
print(s)
print(len(s))
print(hash("Codegnan"))
print(hash(1))
```

```python
#tuple in set
s={1,2,4,(3,4,4,5,6)}
print(s)
print(len(s))
```

```python
#define empty set
s1={}
s2=set()
print(type(s1),type(s2))
```
