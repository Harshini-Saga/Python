```python
#math Module
print(math.ceil(4.956)) #It returns upper bound value
print(math.floor(4.95)) #It returns lower bound value
print(math.factorial(5)) #it returns factorial of a number
print(math.gcd(10,5)) #it returns gcd of numbers
print(math.lcm(45,5,2)) #it returns lcm of numbers
print(math.log10(2))   #IT returns logarithmic value
print(math.log2(32)) 
print(math.log10(4)) 
print(math.pi) 
```

```python
#random module
import random
print(random.random()) #This function generates the value between 0 to 1
print(random.randint(1,10))   #This function genrates the random number between 1 to 10
print(random.choice(['A',1,2,3,4,5,6,8,9,'H','R','A']))
print(random.choices(['A',1,2,3,4,5,6,8,9,'H','R','A'],k=4))
arr=['A',1,2,3,4,5,6,8,9,'H','R','A']
random.shuffle(arr)
print(arr)
```

```python
import random
steps=0
rolls=0
while True:
    input("To roll dice click enter")
    dice=random.randint(1,6)
    rolls+=1
    print(f"Current dice value:{dice}")
    if steps + dice <= 100:
        steps += dice
    if steps==100:
        print(f"You reached top in {rolls} rolls")
        break
    print(f"For {rolls} rolls you are at {steps} position") 
```
