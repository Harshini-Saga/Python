```python
#A.Summation
n=int(input())
nums=list(map(int,input().split()))
sum1=sum(nums)
print(abs(sum1))
```

```python
#B.Searching
n=int(input())
nums=list(map(int,input().split()))
ele=int(input())
if ele in nums:
    print(nums.index(ele))
else:
    print("-1")
```

```python
#C.Replacement
n=int(input())
nums=list(map(int,input().split()))
for i in nums:
    if i>0:
        print("1",end=" ")
    elif i<0:
        print("2",end=" ")
    else:
        print("0",end=" ")
```

```python
#D.Positions in an Array
n = int(input())
a = list(map(int, input().split()))

for i, value in enumerate(a):
    if value <= 10:
        print(f"A[{i}] = {value}")
```

```python
#E.Lowest Number
n=int(input())
a=list(map(int,input().split()))
lower=min(a)
ind=a.index(lower)+1
print(lower,ind)
```

```python
#F.reversing
n=int(input())
lst=list(map(int,input().split()))
for i in lst[::-1]:
    print(i,end=" ")    
```

```python
#G.Palindrome Array
n=int(input())
lst=list(map(int,input().split()))
n1=lst[::-1]
if n1==lst:
    print("YES")
else:
    print("NO")
```

```python
#H.Sorting
n=int(input())
lst=list(map(int,input().split()))
s=sorted(lst)
for i in s:
    print(i,end=" ")
```
