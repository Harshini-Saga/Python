```python
#I.Smallest Pair
t = int(input())

for _ in range(t):
    n = int(input())
    arr = list(map(int, input().split()))

    ans = float('inf')

    for i in range(n):
        for j in range(i + 1, n):
            value = arr[i] + arr[j] + (j - i)
            if value < ans:
                ans = value

    print(ans)
```

```python
#L.SubArray
t = int(input())

for _ in range(t):
    n = int(input())
    arr = list(map(int, input().split()))

    result = []

    for i in range(n):
        current_max = arr[i]

        for j in range(i, n):
            if arr[j] > current_max:
                current_max = arr[j]

            result.append(str(current_max))

    print(" ".join(result))
```

```python
#M.Replace MIN MAX
n=int(input())
l=list(map(int,input().split()))
mini=l.index(min(l))
maxi=l.index(max(l))
l[mini],l[maxi]=l[maxi],l[mini]
print(*l)
```

```python
#O.Fibonacci
def fib(n):
    if n==1:
        return 0
    if n==2:
        return 1
    return fib(n-1)+fib(n-2)
n=int(input())
print(fib(n))
```

```python
#P.Minimize number
n = int(input())
l = list(map(int, input().split()))

ans = float('inf')

for i in l:
    count = 0
    while i % 2 == 0:
        count += 1
        i //= 2
    ans = min(ans, count)

print(ans)
```
