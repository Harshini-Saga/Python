```python
import requests
from bs4 import BeautifulSoup as bs
```

```python
search_ele = "Iphone".replace(" ","")
url="https://www.flipkart.com/search?q="+search_ele
url
```

```python
headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept-Language": "en-US,en;q=0.9",
    "Referer": "https://google.com"
}
content = requests.get(url,headers = headers)
content
```

```python
print(content.text)
```

```python
web_page = bs(content.text,'html.parser')
web_page
```

```python
products = web_page.find_all('div',{'class':"lvJbLV col-12-12"})
print(len(products))
```
