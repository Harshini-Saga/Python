```python
#3.even,odd,positive,negative
n=int(input())
nums=list(map(int,input().split()))
even=0
odd=0
pos=0
neg=0
for i in nums:
    if i%2==0:
        even+=1
    if i%2!=0:
        odd+=1
    if i>0:
        pos+=1
    if i<0:
        neg+=1
print(f"Even: {even}")
print(f"Odd: {odd}")
print(f"Positive: {pos}")
print(f"Negative: {neg}")
```

```python
#I.Palindrome
n = int(input())
s = str(n)
r = s[::-1]
print(int(r))   # removes leading zeros
if s == r:
    print("YES")
else:
    print("NO")
```

```python
#L.GCD
a,b=map(int,input().split())
l=[]
for i in range(1,min(a,b)+1):
    if a%i==0 and b%i==0:
        l.append(i)
print(max(l))
```

```python
#M.Lucky numbers
a,b=map(int,input().split())
found=False
for i in range(a,b+1):
    for j in str(i):
        if j!='4' and j!='7':
            break
    else:
        print(i,end=" ")
        found=True
if not found:
    print(-1)
```

```python
#R.sequence and sum
while True:
    n1,n2=map(int,input().split())
    if n1<=0 or n2<=0:
        break
    sum=0
    start=min(n1,n2)
    end=max(n1,n2)
    for i in range(start,end+1):
        print(i,end=" ")
        sum=sum+i
    print("sum ="+str(sum))
```

```python
#S.Sum of Consecutive Odd numbers
t=int(input())
for i in range(t):
    n1,n2=map(int,input().split())
    sum=0
    start=min(n1,n2)
    end=max(n1,n2)
    for i in range(start+1,end):
        if i%2!=0:
            sum=sum+i
    print(sum)
```

```python
#T.Shape2
n = int(input())

for i in range(n):
    spaces = n - i - 1
    stars = 2 * i + 1
    print(" " * spaces + "*" * stars)
```

```python
#U.Some Sums
n, a, b = map(int, input().split())

total = 0

for i in range(1, n + 1):
    digit_sum = 0

    for digit in str(i):
        digit_sum += int(digit)

    if a <= digit_sum <= b:
        total += i

print(total)
```

```python
#V.PUM
n = int(input())

num = 1

for i in range(n):
    print(num, num + 1, num + 2, "PUM")
    num += 4
```

```python
#W.Shape3
n = int(input())

# Upper half
for i in range(n):
    spaces = n - i - 1
    stars = 2 * i + 1
    print(" " * spaces + "*" * stars)

# Lower half
for i in range(n):
    spaces = i
    stars = 2 * (n - i) - 1
    print(" " * spaces + "*" * stars)
```

```python
#X.Decinal
t = int(input())

for _ in range(t):
    n = int(input())

    ones = 0

    while n > 0:
        ones += n % 2
        n //= 2

    print((2 ** ones) - 1)
```
