```python
#Count vowels and consonants in string
s=input().lower()
v=c=0
for i in s:
    if i.isalpha():
        if i in 'aeiou':
            v +=1
        else:
            c +=1
print(v,c,end=" ")
```

```python
#Count even numbers and odd numbers in a given list of numbers
l=list(map(int,input().split()))
e=o=0
for i in l:
    if i % 2==0:
        e+=1
    else:
        o+=1
print(f"Number of odd are:{o}")
print(f"Number of even are:{e}")
```
