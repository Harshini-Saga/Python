```python
#With parameter with return
#Sum of two numbers
#function defination
def add(x:int,y:int)->int:#it defines what data type is indicate like -> int here
    c=x+y
    return c
#function calling
a,b=5,6
res=add(a,b)
print(res)
```

```python
#With return and without parameter
#Sum of two numbers
#function defination
def add()->int:
    x,y=10,20
    c=x+y
    return c
#function calling
res=add()
print(res)
```

```python
#When function not returns any value or function doesnt contains return statement in this case,function by defaulty it returns "none" as result to function call block
#Without return with parameter
#Sum of two numbers
#function defination
def add(x:int,y:int)->int:
    c=x+y
    print(f"This is from function defination block:{c}")
#function calling
res=add(4,5)
print(f"This is from function calling block:{res}")
```

```python
#Without return and without parameters
def add():
    x,y=10,30
    c=x+y
    print(f"This is from function defination block: {c}")
#function calling
res=add()
print(f"This is from function calling block: {res}")
```

```python
#Find whether the num is prime or not
def prime_check(n:int)->bool:
    if n<2:
        return False
    for i in range(2,n//2+1):
        if n%i==0:
            return False
    return True
num=int(input())
print(prime_check(num))
```
