```python
#Dictionary
#Dictionary Ordered Collection
details ={'name':'Ravi','Age':20,'Course':'Python'}
print(details)
```

```python
#Dictionary cant allow list as key
details ={'name':'Ravi','Age':20,'Course':'Python',[1,2,3]:10}
print(details)
```

```python
#Dictionary allow list as value
#If dictionary contains multiple items with same key, In that case it updates first item key value with last defined item value
details ={'name':'Ravi','age':20,'Course':'Python','marks':[85,90,86],'age':50}
print(details)
```

```python
#Access Values by using keys
#Dictionary Ordered Collection
details ={'name':'Ravi','age':20,'Course':'Python','marks':[85,90,86],'age':50}
print(details['Course'])
print(details['age'])
#Updating the dictionary value
details['age']+=1
details['Course']='java'#It updates value for 'Course' key
print(details)
#item assignment
details['location']="India"#It adds new item to dictionary
print(details)
```

### Dictionary methods
-Get value
    -get(key, default_val):It returns the value if key exists in dictionary otherwise 'default_val'
-Update values and add items to dictionary
    -update(new_dictionary):Dictionary updates with New_dictionary items
-Remove items from dictionary
    -pop(key): It returns value and remove items if key is exists in dictionary
otherwise 'Keyerror'
    -popitem():It returns tuple of key value pair and removes the last item from the dictionary
-Get Keys,value and Items from dictionary
    -keys():It returns list of dictionary keys
    -values():It returns list of dictionary values
    items():It returns list of tuple key-value pairs

```python
details ={'name':'Ravi','age':20,'Course':'Python','marks':[85,90,86],'age':50}
print(details.get('name'))
print(details.get('location'))
print(details.get('age',18))
print(details.get('location',"location key not exists"))
```

```python
details ={'name':'Ravi','age':20,'Course':'Python'}
d={'marks':[85,90,86],'age':50}
details.update(d)
print(details)
```

```python
details ={'name':'Ravi','age':20,'Course':'Python','marks':[85,90,86],'age':50}
print(details)
print(details.pop("marks"))
print(details.popitem())
print(details)
#print(details.pop(percentage))#KeyError:percentage
```
